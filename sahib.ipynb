```python
import numpy as np
# creating a 3D array
A = np.array([[4,5,6],
            [8,9,3],
            [1,3,5]])
B=np.array([[8,5,2],
           [9,6,3],
           [1,4,7]])
print("Matrix A")
print(A)
print("Matrix B")
print(B)
print("Addition")
c=np.add(A,B)
print(c);
print("Mul")
c=np.dot(A,B)
print(c);
```
