```python
# Scenario: A stock trading bot must process 1 million price ticks in under 0.5 seconds to execute a buy/sell decision 
# before the market moves — you use time.time() to benchmark your NumPy calculations and optimize if it's too slow.
# Below is the typical example of how shares are buy/sell
import numpy as np
import time

# Simulate 1 million price ticks/fake price
prices = np.random.rand(1_000_000) * 100

start = time.time()# it records current time in seconds just before the calculation starts
moving_avg = np.mean(prices[-100:])  # Last 100 ticks/prices
decision = "BUY" if moving_avg < 50 else "SELL"
elapsed = time.time() - start

print(f"Decision: {decision}, Time: {elapsed:.5f}s")
if elapsed > 0.5:
    print("Too slow — optimize or upgrade hardware")
```
